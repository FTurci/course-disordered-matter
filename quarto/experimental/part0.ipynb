# Preparing Your Computer for the Experimental Interlude

Our experimental interlude includes running Python scripts to calculate functions and material features described in the lecture. Before continuing, please make sure that you have a working Python installation and that you can successfully run the required code.

## Option 1: Install Anaconda (Recommended)

The easiest way to obtain a complete scientific Python environment is to install **Anaconda**, which includes Python and most of the scientific libraries we will need.

Download and install Anaconda from:

https://www.anaconda.com/download

Make sure to install the version corresponding to your operating system (Windows, macOS, or Linux).

---

## Option 2: Install Spyder

Spyder is a user-friendly scientific Python development environment that resembles MATLAB and is particularly well suited for beginners.

If you installed Anaconda, Spyder is usually included automatically.

To launch Spyder:

- Open **Anaconda Navigator** and click **Launch** next to Spyder, or
- Open a terminal (or Anaconda Prompt) and type:

```bash
spyder
```

If Spyder is not installed, it can be installed through Anaconda Prompt:

```bash
conda install spyder
```

---

## Verifying the Required Libraries

Open Spyder and create a new Python script containing:

```python
import numpy as np
import matplotlib.pyplot as plt

from tqdm import tqdm  # Progress bar for long calculations

print("Everything is working!")
```

Run the script.

If the installation is working correctly, you should see:

```text
Everything is working!
```

and no error messages should appear.

---

## Installing Missing Packages

If you receive an error such as

```text
ModuleNotFoundError: No module named 'tqdm'
```

open an Anaconda Prompt (or terminal) and install the missing package:

```bash
conda install tqdm
```

or

```bash
pip install tqdm
```
